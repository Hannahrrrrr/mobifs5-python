**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    Otzu =  np.full(256, np.inf)
    for T in range(1, 256): 
        n = img8[img8 < T]
        m = img8[img8 >= T]
        if len(n) == 0 or len(m) == 0: #wert wird nicht überschrieben
            continue
        Otzu[T] = len(n) * np.var(n) + len(m) * np.var(m)
    To = np.argmin(Otzu)
    seg = img8.copy()
    seg[seg < To] = 0
    
    return seg, To
```

```python
To = otsu_thresholding(imgf)
print(To)
plt.figure()
```

```python
def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool
    
    dice = 2 * (bin1* bin2).sum() / (bin1).sum() + (bin2).sum()
    
    return dice
```
